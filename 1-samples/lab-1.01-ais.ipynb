Select following elements from the dataset:

- the second row (ids = 2)

```python
# the second row (ids = 2)
print(ais.iloc[2])
print("*"*50)
# rows 4 to 6 (ids = 5 to 7)
print(ais.iloc[5:7])
print("*"*50)
# Columns 6 to 8 (ferr, bmi, ssf)
print(ais.iloc[:, 5:8])
print("*"*50)
# the variable pcBfat (by name!)
print(ais['pcBfat'])
print("*"*50)
# all observations for the sport "Netball"
print(ais[ais.sport == 'Netball'])
print("*"*50)
# just the variable wt of the observations for "Netball"
print(ais[ais.sport == 'Netball']['wt'])
print("*"*50)
# which sports are played by athletes with a BMI higher than 26?
print(ais[ais.bmi > 26])
print("*"*50)
# Also, provide a list of the unique values and a frequency table of how often each sport occurs
print(ais['sport'].unique())
print("*"*50)
print(ais['sport'].value_counts())
```
